# Visualization

```python
df = pd.read_csv("/content/drive/MyDrive/Datasets/gym_membership.csv")

df.head()
```

### Matplotlib

1. Bar chart of membership types — `gym['abonoment_type'].value_counts()`

```python
total = df["abonoment_type"].value_counts()

fig, ax = plt.subplots()

ax.bar(total.index, total.values)
ax.set_xlabel("Abonoment Type")
ax.set_ylabel("Count")
ax.set_title("Distribution of Abonoment Types")

plt.tight_layout()
plt.show()
```

2. Histogram of `avg_time_in_gym`

```python
fig, ax = plt.subplots()
ax.hist(df['avg_time_in_gym'], bins=15, color='green', edgecolor='black', alpha=0.7)

ax.set_title('Average Time Spent in Gym', fontweight='bold')
ax.set_xlabel('Average Time (minutes)')
ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

3. Scatter plot of `Age` vs `visit_per_week`

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(df["Age"], df["visit_per_week"], alpha=0.5, color='red')

ax.set_title('Age vs Visits Per Week')
ax.set_xlabel('Age')
ax.set_ylabel('Visits Per Week')

plt.tight_layout()
plt.show()
```

### SNS

1. Box plot — `gender` vs `Age`

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.boxplot(data=df, x="gender", y="Age")
ax.set_title('Gender vs Ages')

plt.tight_layout()
plt.show()
```

2. Count plot — `days_per_week`

```python
fig, ax = plt.subplots(figsize=(15, 8))
sns.countplot(data=df, x="days_per_week")

ax.set_title("Count of Days Per Week", fontweight="bold")
ax.set_xlabel("Days Per Week")
ax.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

3. Bar plot — mean `visit_per_week` per `abonoment_type`

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.barplot(data=df, x="abonoment_type", y="visit_per_week")
ax.set_title('Visits Per Week by Abonoment Type', fontweight="bold")

plt.tight_layout()
plt.show()
```
