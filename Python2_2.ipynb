```python
#H1 Write a function to calculate the sum of numbers from 1 to N.
```

```python
def total_sum(n):
    return (n * (n + 1)) // 2

print(total_sum(20))
```

```python
#H3 Write a function that checks whether a number is even/odd.
```

```python
def even_odd(n):
    if n % 2 == 0:
        print("Even")
    else:
        print("Odd")

even_odd(2)
```

```python
#H7 Write a Python program to calculate the mean of a list of numbers.
```

```python
def calculate_mean(num):
    return sum(num) / len(num)

print(calculate_mean([1, 2, 3, 4, 5]))
```

```python
#H8 Classify Numbers into Even and Odd Lists
```

```python
def Classify(num):
    even = []
    odd  = []
    for i in range(len(num)):
        if num[i] % 2 == 0:
            even.append(num[i])
        else:
            odd.append(num[i])
    
    return even, odd

even, odd = Classify([1, 2, 3, 4, 5])

print("Even values : ", even)
print("Odd values : ", odd)
```

```python
#H9 Write a program to find the key in a dictionary that has the maximum value.
```

```python
a = {'a': 10, 'b': 20, 'c': 15}

ans = max(a, key=a.get)
print(ans)
```

```python
# H10: Write a program to create a dictionary where the keys are numbers from 1 to n and the values are their squares.
```

```python
def squares_dict(n):
    ans = {}
    for i in range(1, n + 1):
        ans[i] = i ** 2
    return ans

print(squares_dict(5))
```
